```python
def minimax(depth, node, maximizing):
    # Base case: if node is a number (leaf)
    if isinstance(node, int):
        return node

    if maximizing:
        best = -999
        for value in node:
            best = max(best, minimax(depth + 1, value, False))
        return best
    else:
        best = 999
        for value in node:
            best = min(best, minimax(depth + 1, value, True))
        return best


# Game tree
tree = [
    [3, 5],
    [2, 9],
    [12, 5],
    [23, 23]
]

result = minimax(0, tree, True)
print("Best value:", result)
```
