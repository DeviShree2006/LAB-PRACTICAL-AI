```python
facts = {
    "Socrates is human",
    "Plato is human"
}

rules = {
    "Socrates is mortal": "Socrates is human",
    "Plato is mortal": "Plato is human"
}


def backward_chaining(goal):

    # Check if goal is directly available
    if goal in facts:
        return True

    # Check rules
    if goal in rules:

        required_fact = rules[goal]

        if required_fact in facts:
            return True

    return False


goal = input("Enter goal: ")

if backward_chaining(goal):
    print("Goal is TRUE")
else:
    print("Goal is FALSE")
```
