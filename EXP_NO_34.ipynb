```python
diet = {
    "diabetes": "Low sugar and high fiber diet",
    "obesity": "Low calorie and low fat diet",
    "anemia": "Iron rich food",
    "hypertension": "Low salt diet"
}

disease = input("Enter disease: ").lower()

if disease in diet:
    print("Suggested diet:", diet[disease])
else:
    print("Disease not found")
```
