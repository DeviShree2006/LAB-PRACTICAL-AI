```python
from collections import deque

def valid(m, c):
    # Missionaries and cannibals on the left side
    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    # Right side
    mr = 3 - m
    cr = 3 - c

    # Cannibals should not outnumber missionaries
    if m > 0 and c > m:
        return False

    if mr > 0 and cr > mr:
        return False

    return True


def solve():
    # (Missionaries_left, Cannibals_left, Boat_position)
    start = (3, 3, 0)
    goal = (0, 0, 1)

    moves = [
        (1, 0), (2, 0),
        (0, 1), (0, 2),
        (1, 1)
    ]

    queue = deque([(start, [start])])
    visited = set()

    while queue:
        state, path = queue.popleft()

        if state in visited:
            continue

        visited.add(state)

        if state == goal:
            return path

        m, c, boat = state

        for dm, dc in moves:

            if boat == 0:
                new_state = (m - dm, c - dc, 1)
            else:
                new_state = (m + dm, c + dc, 0)

            if valid(new_state[0], new_state[1]):
                if new_state not in visited:
                    queue.append((new_state, path + [new_state]))

    return None


solution = solve()

print("Solution:")

for state in solution:
    print(state)
```
