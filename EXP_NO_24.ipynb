```python
def alphabeta(depth, node, alpha, beta, maximizing):

    if depth == 3:
        return node

    if maximizing:

        best = -999

        for value in node:

            best = max(
                best,
                alphabeta(
                    depth + 1,
                    value,
                    alpha,
                    beta,
                    False
                )
            )

            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:

        best = 999

        for value in node:

            best = min(
                best,
                alphabeta(
                    depth + 1,
                    value,
                    alpha,
                    beta,
                    True
                )
            )

            beta = min(beta, best)

            if beta <= alpha:
                break

        return best


tree = [
    [3, 5],
    [2, 9],
    [12, 5],
    [23, 23]
]

result = alphabeta(
    0,
    tree,
    -999,
    999,
    True
)

print("Best value:", result)
```
