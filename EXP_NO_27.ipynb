```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Example dataset
data = pd.DataFrame({
    "feature": [1, 2, 3, 4, 5, 6],
    "label": ["Open", "Close", "Open", "Pending", "Close", "Clo"]  # Notice 'Clo'
})

# Split into train/test
X_train, X_test, y_train, y_test = train_test_split(
    data[["feature"]], data["label"], test_size=0.3, random_state=42
)

# Initialize LabelEncoder
le = LabelEncoder()
le.fit(y_train)  # Fit only on training labels

# Handle unseen labels in test set
# Replace unseen labels with "Unknown"
y_test_cleaned = [
    label if label in le.classes_ else "Unknown" for label in y_test
]

# Extend encoder classes to include "Unknown"
if "Unknown" not in le.classes_:
    le.classes_ = np.append(le.classes_, "Unknown")

# Transform labels
y_train_enc = le.transform(y_train)
y_test_enc = le.transform(y_test_cleaned)

# Train a simple model
model = LogisticRegression()
model.fit(X_train, y_train_enc)

# Predict
y_pred = model.predict(X_test)

# Decode predictions back to original labels
y_pred_labels = le.inverse_transform(y_pred)

print("Original test labels:", list(y_test))
print("Cleaned test labels:", list(y_test_cleaned))
print("Predicted labels:", list(y_pred_labels))
```
