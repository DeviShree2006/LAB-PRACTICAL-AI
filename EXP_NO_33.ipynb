```python
female = {"Pam", "Liz", "Ann", "Pat"}
male = {"Tom", "Bob", "Jim"}

parents = {
    "Liz": ["Pam", "Tom"],
    "Ann": ["Pam", "Tom"],
    "Pat": ["Liz", "Bob"],
    "Jim": ["Liz", "Bob"]
}


def mother(x, y):
    return x in female and x in parents.get(y, [])


def father(x, y):
    return x in male and x in parents.get(y, [])


def sister(x, y):
    if x not in female or x == y:
        return False

    return set(parents.get(x, [])) & set(parents.get(y, []))


def brother(x, y):
    if x not in male or x == y:
        return False

    return bool(set(parents.get(x, [])) & set(parents.get(y, [])))


print("Mother of Liz:", [x for x in female if mother(x, "Liz")])
print("Father of Liz:", [x for x in male if father(x, "Liz")])

print("Sisters of Pat:",
      [x for x in female if sister(x, "Pat")])

print("Brothers of Pat:",
      [x for x in male if brother(x, "Pat")])
```
