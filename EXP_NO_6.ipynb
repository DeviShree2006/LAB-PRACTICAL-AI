```python
my_list = []

# Correct way: use append instead of add
my_list.append(1)
print("List after appending an item:", my_list)

my_list.append(2)
print("List after appending another item:", my_list)

my_list.extend([3, 4, 5])
print("List after extending using extend method:", my_list)

my_list.remove(3)
print("List after removing an item using remove method:", my_list)
```
