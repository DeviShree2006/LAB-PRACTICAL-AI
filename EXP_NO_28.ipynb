```python
from sklearn.neural_network import MLPClassifier

# Training data
X = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

# AND operation
y = [0, 0, 0, 1]

# Feed Forward Neural Network
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='relu',
    max_iter=2000,
    random_state=1
)

model.fit(X, y)

# Test
result = model.predict([[1, 1]])

print("Prediction:", result[0])
```
