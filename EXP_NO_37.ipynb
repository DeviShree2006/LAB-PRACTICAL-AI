```python
symptoms = {
    "Ravi": ["fever", "cough", "body pain"],
    "Priya": ["fever", "headache", "cold"]
}


def diagnose(person):

    s = symptoms.get(person, [])

    if "fever" in s and "cough" in s and "body pain" in s:
        return "Flu"

    elif "fever" in s and "headache" in s and "cold" in s:
        return "Cold"

    else:
        return "Unknown"


person = input("Enter patient name: ").title()

print("Possible disease:", diagnose(person))
```
