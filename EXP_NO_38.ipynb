```python
facts = {
    "Socrates is human",
    "Plato is human"
}

rules = {
    "Socrates is human": "Socrates is mortal",
    "Plato is human": "Plato is mortal"
}


def forward_chaining():

    new_facts = set(facts)

    for condition, conclusion in rules.items():

        if condition in new_facts:
            new_facts.add(conclusion)

    return new_facts


result = forward_chaining()

print("Derived Facts:")

for fact in result:
    print(fact)
```
