```python
from collections import deque

def water_jug():
    capacity1 = 4
    capacity2 = 3
    goal = 2

    queue = deque()
    queue.append((0, 0, []))

    visited = set()

    while queue:
        a, b, path = queue.popleft()

        if (a, b) in visited:
            continue

        visited.add((a, b))

        path = path + [(a, b)]

        if a == goal or b == goal:
            return path

        states = []

        # Fill Jug 1
        states.append((capacity1, b))

        # Fill Jug 2
        states.append((a, capacity2))

        # Empty Jug 1
        states.append((0, b))

        # Empty Jug 2
        states.append((a, 0))

        # Pour Jug 1 -> Jug 2
        amount = min(a, capacity2 - b)
        states.append((a - amount, b + amount))

        # Pour Jug 2 -> Jug 1
        amount = min(b, capacity1 - a)
        states.append((a + amount, b - amount))

        for state in states:
            if state not in visited:
                queue.append((state[0], state[1], path))

    return None


solution = water_jug()

if solution:
    print("Solution:")
    for state in solution:
        print("Jug 1:", state[0], "litres",
              "| Jug 2:", state[1], "litres")
else:
    print("No solution")
```
