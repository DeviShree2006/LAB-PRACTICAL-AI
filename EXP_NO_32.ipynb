```python
birds = {
    "Parrot": True,
    "Eagle": True,
    "Penguin": False,
    "Ostrich": False
}

bird = input("Enter bird name: ").title()

if bird in birds:

    if birds[bird]:
        print(bird, "can fly")
    else:
        print(bird, "cannot fly")

else:
    print("Bird not found")
```
