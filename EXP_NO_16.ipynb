```python
def vacuum_cleaner():

    rooms = {
        "A": "Dirty",
        "B": "Dirty"
    }

    location = "A"

    while rooms["A"] == "Dirty" or rooms["B"] == "Dirty":

        print("Vacuum is in room", location)

        if rooms[location] == "Dirty":
            print("Cleaning room", location)
            rooms[location] = "Clean"

        else:
            if location == "A":
                print("Moving to room B")
                location = "B"
            else:
                print("Moving to room A")

                location = "A"

    print("Both rooms are clean.")


vacuum_cleaner()
```
