```python
def monkey_banana():

    monkey_position = "floor"
    monkey_under_box = True
    monkey_on_box = False
    banana = False

    print("Initial State:")
    print("Monkey is on the floor")

    if monkey_position == "floor" and monkey_under_box:
        print("Monkey pushes the box")
        print("Monkey climbs the box")

        monkey_on_box = True
        monkey_position = "box"

    if monkey_on_box:
        print("Monkey reaches the banana")
        banana = True

    if banana:
        print("Monkey gets the banana")


monkey_banana()
```
