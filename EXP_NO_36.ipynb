```python
fruits = [
    ("Apple", "Red"),
    ("Banana", "Yellow"),
    ("Orange", "Orange"),
    ("Grapes", "Green"),
    ("Mango", "Yellow"),
    ("Strawberry", "Red")
]

color = input("Enter color: ").title()

print("Fruits with", color, "color:")

found = False

for fruit, fruit_color in fruits:

    if fruit_color == color:
        print(fruit)
        found = True

if not found:
    print("No fruit found")
```
